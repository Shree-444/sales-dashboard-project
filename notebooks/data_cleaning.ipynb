```python
import pandas as pd

original_df = pd.read_csv("../data/raw/superstore.csv", encoding="latin1")

# creating a copy of the original dataset
work_df = original_df.copy()

# dropping rows with NA values
work_df = work_df.dropna(axis = 0)

# standardizing column names as "column_name"
work_df.columns = (work_df.columns
                    .str.strip()
                    .str.lower()
                    .str.replace(" ", "_")
)

# changing datatype of dates from string to date-time
work_df["order_date"] = pd.to_datetime(work_df["order_date"])
work_df["ship_date"] = pd.to_datetime(work_df["ship_date"])

# creating new derived columns for future usage
work_df["year"] = work_df["order_date"].dt.year
work_df["month"] = work_df["order_date"].dt.month_name()
work_df["month_number"] = work_df["order_date"].dt.month
work_df["month_year"] = work_df["order_date"].dt.to_period("M")
work_df["profit_margin"] = (work_df["profit"] / work_df["sales"]) * 100

# saving the file to processed data folder
work_df.to_csv("../data/processed/processed_superstore.csv", index=False)

```
